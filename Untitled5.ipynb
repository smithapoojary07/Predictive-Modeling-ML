```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import os

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report,
    ConfusionMatrixDisplay, roc_curve, auc
)

os.makedirs("ml_results", exist_ok=True)

print("Libraries imported successfully!")
```

```python
dataset = load_breast_cancer(as_frame=True)

X = dataset.data
y = dataset.target

print("Dataset loaded successfully!")
print("Number of records:", X.shape[0])
print("Number of features:", X.shape[1])
print("Target classes:", list(dataset.target_names))

display(X.head())
```

```python
print("Dataset information:")
display(X.describe().T)

print("\nMissing values:")
display(X.isnull().sum().to_frame("Missing Values"))

print("\nTarget class distribution:")
print(y.value_counts().rename(index={
    0: "Malignant",
    1: "Benign"
}))

print("\nDuplicate rows:", X.duplicated().sum())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
models = {
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    )
}

results = []
predictions = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    predictions[name] = y_pred

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1 Score": f1_score(y_test, y_pred)
    })

results_df = pd.DataFrame(results)

print("Model training completed!")
display(results_df.round(4))
```

```python
results_df.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1 Score"]
].plot(kind="bar", figsize=(10, 6), ylim=(0, 1.05))

plt.title("Machine Learning Model Performance Comparison")
plt.ylabel("Score")
plt.xlabel("Model")
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.tight_layout()

plt.savefig("ml_results/model_comparison.png", dpi=200)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax, (name, model) in zip(axes, models.items()):
    ConfusionMatrixDisplay.from_predictions(
        y_test,
        predictions[name],
        display_labels=dataset.target_names,
        cmap="Blues",
        ax=ax
    )
    ax.set_title(name)

plt.tight_layout()
plt.savefig("ml_results/confusion_matrices.png", dpi=200)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

for name, model in models.items():
    positive_index = list(model.classes_).index(0)
    probabilities = model.predict_proba(X_test)[:, positive_index]

    fpr, tpr, thresholds = roc_curve(
        (y_test == 0).astype(int),
        probabilities
    )

    roc_auc = auc(fpr, tpr)

    plt.plot(
        fpr, tpr,
        label=f"{name} (AUC = {roc_auc:.3f})"
    )

plt.plot([0, 1], [0, 1], "k--", label="Random baseline")
plt.title("ROC Curve Comparison")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.tight_layout()

plt.savefig("ml_results/roc_curve.png", dpi=200)
plt.show()
```

```python
best_row = results_df.loc[results_df["F1 Score"].idxmax()]
best_model_name = best_row["Model"]
best_model = models[best_model_name]

joblib.dump(best_model, "ml_results/best_model.pkl")
results_df.to_csv("ml_results/model_results.csv", index=False)

with open("ml_results/classification_report.txt", "w") as file:
    file.write(f"Selected model: {best_model_name}\n\n")
    file.write(
        classification_report(
            y_test,
            predictions[best_model_name],
            target_names=dataset.target_names
        )
    )

print("Selected model:", best_model_name)
print("Model and evaluation results saved successfully!")
```

```python
sample = X_test.iloc[[0]]

predicted_class = best_model.predict(sample)[0]
predicted_label = dataset.target_names[predicted_class]

print("Predicted class:", predicted_label)
print("Actual class:", dataset.target_names[y_test.iloc[0]])
```

```python
from IPython.display import display, Image

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Model comparison
results_df.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1 Score"]
].plot(kind="bar", ax=axes[0], ylim=(0, 1.05))
axes[0].set_title("Model Comparison")
axes[0].tick_params(axis="x", rotation=20)

# Confusion matrix for the selected model
ConfusionMatrixDisplay.from_predictions(
    y_test,
    predictions[best_model_name],
    display_labels=dataset.target_names,
    cmap="Blues",
    ax=axes[1]
)
axes[1].set_title(f"Confusion Matrix: {best_model_name}")

# Feature importance for tree-based models
importances = pd.Series(
    best_model.feature_importances_,
    index=X.columns
).nlargest(10).sort_values()

importances.plot(kind="barh", ax=axes[2])
axes[2].set_title("Top 10 Important Features")

plt.suptitle("Predictive Modeling Dashboard", fontsize=16)
plt.tight_layout()
plt.savefig("ml_results/all_visualizations.png", dpi=200)
plt.show()
```
